```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('train.csv')
```

```python
data = sns.load_dataset("iris")
```

```python
#relational plot
#line plot
#scatter plot
plt.figure(figsize=(10,6))
sns.lineplot(x=["sepal_length"], y=["sepal_width"])
sns.relplot(x="sepal_length", y="sepal_width", data=data, hue="species", style="species", size="petal_length")

```

```python
plt.Figure(figsize=(3, 3))
sns.scatterplot(x=data['sepal_length'],y=data['petal_length'], hue=data['species'], style=data['species'])
plt.show()
```

```python
#distribution plots:
#To analyze the distribution of a single variable, we can use distribution plots. These plots help us understand the underlying distribution of the data, such as whether it is normal, skewed, or has multiple peaks.
# dist plot/hist plot:
sns.histplot(data=data, x='sepal_length', kde=True)
# box plot: to see the outliers in the data
#sns.boxplot(x=data['species'], y=data['sepal_length'])
```

```python
plt.figure(figsize=(3,3))
sns.displot(x=data['sepal_length'])
plt.show()
```

```python
# box plot
sns.boxplot(x=data['petal_length'])
```

```python
#count plot: to see the count of each category in a categorical variable
sns.countplot(x=data['species'])
```

```python
#bar plot: to see the relationship between a categorical variable and a numerical variable
sns.barplot(x=data['species'], y=data['petal_width'])
```

```python
#box plot
sns.boxplot(x=data['species'], y=data['sepal_length'], hue=data['species'])
```

```python
#violin plot: combination of histoplot and distplot
sns.violinplot(x=data['species'], y=data['petal_length'], hue=data['species'])
```

```python
#strip plot
sns.stripplot(x=data['species'], y=data['petal_length'], hue=data['species'])
```

```python
#swarmplot
sns.swarmplot(x=data['species'], y=data['petal_length'], hue=data['species'])
```
