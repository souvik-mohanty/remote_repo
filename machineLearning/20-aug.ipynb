# K-Means Clustering

## 1. Definition

**K-Means Clustering** is an **unsupervised machine learning algorithm** used to divide a dataset into **K distinct clusters**.

The algorithm groups data points such that:

* Data points within the **same cluster are as similar as possible**.
* Data points belonging to **different clusters are as different as possible**.

K-Means primarily uses **distance**, usually **Euclidean distance**, to determine similarity.

The main objective is to minimize the **Within-Cluster Sum of Squares (WCSS)**, also called **inertia**.

---

## 2. How K-Means Works

Suppose we have a dataset and want to divide it into **K = 3 clusters**.

### Step 1: Choose K

Select the number of clusters, `K`.

Example:

```text
K = 3
```

The value of `K` is usually determined using techniques such as:

* Elbow Method
* Silhouette Score

---

### Step 2: Initialize Centroids

Select `K` initial points as **centroids**.

A centroid represents the center of a cluster.

The initialization can be:

* Random initialization
* **K-Means++** initialization

> **K-Means++ is generally preferred** because it selects better-spread initial centroids and reduces the chance of poor clustering.

---

### Step 3: Assignment Step

Calculate the distance between every data point and each centroid.

Each point is assigned to the **nearest centroid**.

For Euclidean distance:

[
d(x,c)=\sqrt{\sum_{i=1}^{n}(x_i-c_i)^2}
]

where:

* `x` = data point
* `c` = centroid
* `n` = number of features

---

### Step 4: Update Step

After assigning all points to clusters, calculate a new centroid for each cluster.

The centroid is the **mean of all points belonging to that cluster**.

For a cluster containing (n) points:

[
C = \frac{1}{n}\sum_{i=1}^{n}x_i
]

---

### Step 5: Repeat

Repeat the following two steps:

```text
Assignment → Update → Assignment → Update → ...
```

The process stops when:

* Centroids stop changing significantly, or
* The decrease in inertia becomes very small, or
* The maximum number of iterations is reached.

---

# 3. Important Terminology

## Centroid

A **centroid** is the mathematical center of a cluster.

It is calculated as the mean of all data points assigned to that cluster.

Example:

```text
Points: 2, 4, 6

Centroid = (2 + 4 + 6) / 3
         = 4
```

---

## Inertia / WCSS

**Inertia**, also called **Within-Cluster Sum of Squares (WCSS)**, measures how close the data points are to their assigned centroid.

[
WCSS = \sum_{j=1}^{K}\sum_{x_i \in C_j}
|x_i-\mu_j|^2
]

where:

* (K) = number of clusters
* (C_j) = cluster `j`
* (x_i) = data point
* (\mu_j) = centroid of cluster `j`

### Goal of K-Means

[
\boxed{\text{Minimize WCSS / Inertia}}
]

Lower inertia means points are, overall, closer to their cluster centers.

However, **inertia will generally decrease as K increases**, so the lowest inertia alone cannot be used to select the best K.

---

# 4. Elbow Method

The **Elbow Method** is commonly used to estimate the optimal value of `K`.

### Procedure

1. Train K-Means for different values of `K`.
2. Calculate inertia for each `K`.
3. Plot:

```text
K → Inertia
```

4. Find the point where the reduction in inertia starts becoming much smaller.

This point is called the **elbow**.

### Example

```text
Inertia
  |
  |\
  | \
  |  \
  |   \__
  |      \__
  |         \__
  +---------------- K
        ↑
      Elbow
```

The elbow represents a reasonable trade-off between:

* Number of clusters
* Reduction in within-cluster variation

---

# 5. Silhouette Score

The **Silhouette Score** evaluates how well each data point fits within its assigned cluster compared with other clusters.

Its value ranges from:

[
-1 \leq S \leq 1
]

### Interpretation

| Score         | Meaning                                    |
| ------------- | ------------------------------------------ |
| Close to `1`  | Well-separated and cohesive clusters       |
| Around `0`    | Overlapping clusters                       |
| Less than `0` | Point may be assigned to the wrong cluster |

A **higher average Silhouette Score** generally indicates better-defined clusters.

---

# 6. Feature Scaling

Feature scaling is **very important** for K-Means because K-Means is distance-based.

Suppose we have:

```text
Age:     20 – 60
Salary:  20,000 – 2,00,000
```

Salary has a much larger numerical scale.

Without scaling, salary may dominate the distance calculation.

Common scaling techniques:

### Standardization

[
z = \frac{x-\mu}{\sigma}
]

Usually implemented using:

```python
StandardScaler()
```

### Normalization

Scales values to a particular range, commonly `[0, 1]`.

```python
MinMaxScaler()
```

> **Interview point:** Feature scaling is generally required when features have significantly different scales.

---

# 7. K-Means++ Initialization

Randomly selecting initial centroids can produce poor clustering results.

**K-Means++** improves initialization by choosing initial centroids that are well separated from each other.

### Benefits

* Better initial centroids
* Usually faster convergence
* Lower chance of poor local solutions
* More stable results than simple random initialization

In Scikit-learn, K-Means uses **K-Means++ initialization by default**.

# 9. Limitations

## 1. Requires K in Advance

K-Means requires the number of clusters to be specified beforehand.

Methods such as:

* Elbow Method
* Silhouette Score

can help estimate an appropriate value.

---

## 2. Sensitive to Outliers

Extreme data points can significantly move the centroid because the centroid is calculated using the **mean**.

---

## 3. Sensitive to Initialization

Different initial centroids can sometimes produce different results.

### Solution

Use:

```text
K-Means++
```

and/or run the algorithm multiple times with different initializations.

---

## 4. Assumes Suitable Cluster Shapes

K-Means works best when clusters are approximately:

* Spherical
* Compact
* Well separated

It may perform poorly when clusters are:

* Irregularly shaped
* Highly overlapping
* Highly elongated
* Very different in density or size

---

## 5. Distance-Based

Because clustering depends on distance, inappropriate feature scaling can produce misleading clusters.

---

# 10. Common Applications

### Customer Segmentation

Group customers based on:

* Spending
* Age
* Purchase frequency
* Income

### Image Compression

Group similar colors together and represent them using cluster centroids.

### Document Clustering

Group documents based on their feature representations.

### Anomaly Detection

Points that are unusually far from their cluster centroids can potentially be treated as anomalies.

> K-Means itself is **not primarily an anomaly-detection algorithm**; it can be used as part of an anomaly-detection approach.

---

# 11. K-Means Objective

The mathematical objective of K-Means is:

[
\boxed{
\min \sum_{j=1}^{K}\sum_{x_i\in C_j}
|x_i-\mu_j|^2
}
]

In simple terms:

> **Assign points to clusters and choose centroids so that the total squared distance between points and their assigned centroids is minimized.**

---

# 12. Complete Algorithm

```text
Choose K
   ↓
Initialize K centroids
   ↓
Calculate distance from every point to each centroid
   ↓
Assign each point to nearest centroid
   ↓
Calculate new centroids
   ↓
Check convergence
   ↓
If not converged → Repeat
   ↓
Final clusters
```

---

# 13. Important Interview Points

* K-Means is an **unsupervised learning algorithm**.
* It is mainly used for **clustering**.
* `K` represents the **number of clusters**.
* A **centroid is the mean of points in a cluster**.
* K-Means tries to minimize **WCSS/Inertia**.
* **Elbow Method** helps estimate `K`.
* **Silhouette Score** evaluates cluster quality.
* Feature scaling is important because K-Means is **distance-based**.
* K-Means is sensitive to **outliers and initialization**.
* **K-Means++** provides better centroid initialization.
* K-Means works best for **compact, roughly spherical clusters**.
* Increasing `K` generally decreases inertia, so **minimum inertia alone does not determine the optimal K**.

---

## 14. One-Line Summary

> **K-Means repeatedly assigns each data point to its nearest centroid and then updates the centroids using the mean of their assigned points until the clusters stabilize.**



# K-Means Clustering

**K-Means Clustering** is an **unsupervised machine learning algorithm**
used to divide a dataset into **K distinct clusters**.

## How K-Means Works

1. **Choose K** — Select the number of clusters.
2. **Initialize Centroids** — Select K initial centroids.
3. **Assignment Step** — Assign each data point to the nearest centroid.
4. **Update Step** — Calculate the mean of each cluster.
5. **Repeat** — Continue until the centroids stabilize.
