# Anonymous Function
* they are also called as lambda functions
* They are one line funtion without any name
* They will have only one expression and any number of parameters

```python
addition=lambda a,b:a+b
addition(a,b)

addition(4,1)
```

```python
greater=lambda a,b:a if a>b else b
```

# Filter()

* it is applicable on iteritable elements like list, string


```python
list1=[10,5,8,14,-5,-6]
negative=filter(lambda a:a<0,list1)
print(list(negative))
```
