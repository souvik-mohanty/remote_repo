# 3->Set

* collection of elements inside {}
* It is a mutable data structure
* Indexing is not possible on set elements
* Duplicates are ignored

```python
# ==========================================================
#                 PYTHON SETS - COMPLETE NOTES
# ==========================================================


# ==========================================================
# What is a Set?
# ==========================================================
# A set is an unordered collection of unique elements.
#
# Important Points:
# ✔ Unordered (no indexing)
# ✔ Mutable (can add/remove elements)
# ✔ Duplicate values are not allowed
# ✔ Stores immutable objects only
# ✔ Created using {}

# ==========================================================
# Creating a Set
# ==========================================================

numbers = {1, 2, 3, 4}
print(numbers)

# Duplicate values are removed automatically
data = {1, 2, 2, 3, 3, 4}
print(data)

# Empty set
empty = set()
print(type(empty))

# ==========================================================
# add()
# ==========================================================

fruits = {"Apple", "Banana"}
# Add one element
fruits.add("Orange")
print(fruits)

# ==========================================================
# update()
# ==========================================================

fruits = {"Apple", "Banana"}
# Add multiple elements
fruits.update(["Mango", "Grapes"])
print(fruits)

# ==========================================================
# remove()
# ==========================================================

fruits = {"Apple", "Banana", "Orange"}
# Removes element
fruits.remove("Banana")
print(fruits)

# remove() raises KeyError if item doesn't exist
# fruits.remove("Kiwi")

# ==========================================================
# discard()
# ==========================================================

fruits = {"Apple", "Banana"}

# Does not raise error if element is absent
fruits.discard("Kiwi")
print(fruits)

# ==========================================================
# pop()
# ==========================================================

colors = {"Red", "Green", "Blue"}

# Removes a random element
removed = colors.pop()

print("Removed:", removed)
print(colors)

# ==========================================================
# clear()
# ==========================================================

numbers = {1,2,3}
numbers.clear()
print(numbers)

# ==========================================================
# copy()
# ==========================================================

a = {1,2,3}
b = a.copy()

print(a)
print(b)

# ==========================================================
# union()
# ==========================================================

A = {1,2,3}
B = {3,4,5}

print(A.union(B))
print(A | B)

# ==========================================================
# intersection()
# ==========================================================

print(A.intersection(B))
print(A & B)

# ==========================================================
# difference()
# ==========================================================

print(A.difference(B))
print(A - B)

# ==========================================================
# symmetric_difference()
# ==========================================================

print(A.symmetric_difference(B))
print(A ^ B)

# ==========================================================
# intersection_update()
# ==========================================================

A = {1,2,3}
B = {2,3,4}

A.intersection_update(B)
print(A)

# ==========================================================
# difference_update()
# ==========================================================

A = {1,2,3}
B = {2}

A.difference_update(B)
print(A)

# ==========================================================
# symmetric_difference_update()
# ==========================================================

A = {1,2,3}
B = {3,4}

A.symmetric_difference_update(B)
print(A)

# ==========================================================
# issubset()
# ==========================================================

A = {1,2}
B = {1,2,3,4}

print(A.issubset(B))

# ==========================================================
# issuperset()
# ==========================================================

print(B.issuperset(A))

# ==========================================================
# isdisjoint()
# ==========================================================

X = {1,2}
Y = {3,4}

print(X.isdisjoint(Y))

# ==========================================================
# Membership Operators
# ==========================================================

nums = {10,20,30}

print(20 in nums)
print(100 not in nums)

# ==========================================================
# Built-in Functions
# ==========================================================

nums = {10,20,30}

print(len(nums))
print(max(nums))
print(min(nums))
print(sum(nums))
print(sorted(nums))
print(any(nums))
print(all(nums))

# ==========================================================
# Converting to Set
# ==========================================================

text = "programming"

# Removes duplicate characters
print(set(text))

# ==========================================================
# Frozen Set
# ==========================================================

# Immutable version of set
fs = frozenset([1,2,3])

print(fs)

# fs.add(4)  # AttributeError

# ==========================================================
# Common Errors
# ==========================================================

# Sets cannot contain mutable objects like lists.
#
# invalid = {[1,2]}
#
# TypeError: unhashable type: 'list'

# ==========================================================
# Time Complexity
# ==========================================================

# add()                     O(1)
# remove()                  O(1)
# discard()                 O(1)
# pop()                     O(1)
# union()                   O(len(A)+len(B))
# intersection()            O(min(len(A),len(B)))
# difference()              O(len(A))
# symmetric_difference()    O(len(A)+len(B))
# in                        O(1)

# ==========================================================
# Interview Questions
# ==========================================================

# Q1. Can a set contain duplicate values?
# No.

# Q2. Can we access set elements using indexing?
# No.

# Q3. Difference between remove() and discard()?
# remove() -> KeyError if missing.
# discard() -> No error.

# Q4. Difference between pop() in list and set?
# List: removes last/indexed element.
# Set : removes a random element.

# Q5. Difference between {} and set()?
# {} -> Empty dictionary
# set() -> Empty set

# ==========================================================
# Quick Revision
# ==========================================================

# add()                       Add one item
# update()                    Add multiple items
# remove()                    Remove item (error if absent)
# discard()                   Remove safely
# pop()                       Remove random item
# clear()                     Remove all items
# copy()                      Copy set
# union() or |                Combine sets
# intersection() or &         Common elements
# difference() or -           Unique elements
# symmetric_difference() or ^ Non-common elements
# issubset()                  Check subset
# issuperset()                Check superset
# isdisjoint()                No common elements

# Important:
# ✔ Unordered
# ✔ No duplicates
# ✔ Mutable
# ✔ No indexing
# ✔ Membership checking is O(1)
```

# 4-> Dictonary
* collection of elements in key value pairs inside {}
* Its  a mutable data structure
* We use keys to acces values

```python

# ==========================================================
#              PYTHON DICTIONARY - COMPLETE NOTES
# ==========================================================

# ----------------------------------------------------------
# What is a Dictionary?
# ----------------------------------------------------------
# A dictionary stores data as key-value pairs.
#
# Important Points:
# ✔ Mutable
# ✔ Unordered (insertion order preserved in Python 3.7+)
# ✔ Keys must be unique and immutable
# ✔ Values can be of any data type
# ✔ Created using {}

# ==========================================================
# Creating Dictionaries
# ==========================================================

student = {
    "name": "Souvik",
    "age": 22,
    "branch": "CSE"
}

print(student)

# Empty dictionary
empty = {}
print(type(empty))

# Using dict()
person = dict(name="Alice", age=25)
print(person)

# ==========================================================
# Accessing Values
# ==========================================================

student = {"name":"Rahul","age":21}

# Access using key
print(student["name"])

# Using get()
print(student.get("age"))

# Returns None if key not found
print(student.get("city"))

# Default value if key not found
print(student.get("city","Not Found"))

# ==========================================================
# Adding & Updating
# ==========================================================

student = {"name":"Rahul"}

# Add new key
student["age"] = 21

# Update value
student["name"] = "Rohan"

print(student)

# update()
student.update({"branch":"CSE","cgpa":8.5})
print(student)

# ==========================================================
# Removing Elements
# ==========================================================

data = {"a":1,"b":2,"c":3}

# pop() removes key and returns value
value = data.pop("b")
print(value)
print(data)

# popitem() removes last inserted pair
pair = data.popitem()
print(pair)

# del keyword
marks = {"math":90,"science":85}
del marks["math"]
print(marks)

# clear()
marks.clear()
print(marks)

# ==========================================================
# Dictionary Methods
# ==========================================================

employee = {"id":101,"name":"John","dept":"IT"}

# keys()
print(employee.keys())

# values()
print(employee.values())

# items()
print(employee.items())

# copy()
emp2 = employee.copy()
print(emp2)

# fromkeys()
keys = ["A","B","C"]
new_dict = dict.fromkeys(keys,0)
print(new_dict)

# setdefault()
employee.setdefault("salary",50000)
print(employee)

# ==========================================================
# Iterating Dictionary
# ==========================================================

student = {"name":"Amit","age":20,"city":"Delhi"}

# Iterate keys
for key in student:
    print(key)

# Iterate values
for value in student.values():
    print(value)

# Iterate key-value pairs
for key, value in student.items():
    print(key,"=",value)

# ==========================================================
# Membership
# ==========================================================

student = {"name":"Rahul","age":20}

print("name" in student)
print("city" not in student)

# ==========================================================
# Built-in Functions
# ==========================================================

marks = {"Math":90,"Science":80,"English":95}

print(len(marks))
print(max(marks))
print(min(marks))
print(sorted(marks))

# Sum values
print(sum(marks.values()))

# ==========================================================
# Nested Dictionary
# ==========================================================

students = {
    101: {"name":"Amit","age":20},
    102: {"name":"Neha","age":21}
}

print(students[101]["name"])

# ==========================================================
# Dictionary Comprehension
# ==========================================================

squares = {x:x*x for x in range(1,6)}
print(squares)

# ==========================================================
# Important Errors
# ==========================================================

# Keys must be immutable.

# invalid = {[1,2]:"List"}
# TypeError: unhashable type: 'list'

# ==========================================================
# Time Complexity
# ==========================================================

# Access by key        O(1)
# Insert               O(1)
# Update               O(1)
# Delete               O(1)
# Search (in keys)     O(1)
# Iteration            O(n)

# ==========================================================
# Interview Questions
# ==========================================================

# Q1. Difference between list and dictionary?
# List stores values by index.
# Dictionary stores key-value pairs.

# Q2. Difference between [] and get()?
# [] -> KeyError if key missing.
# get() -> Returns None/default value.

# Example
d = {"name":"Amit"}
# print(d["age"])      # KeyError
print(d.get("age"))

# Q3. Can dictionary have duplicate keys?
# No. Latest value overwrites previous one.

d = {"a":1,"a":2}
print(d)

# Q4. Can dictionary values be duplicated?
# Yes.

# Q5. Difference between pop() and popitem()?
# pop(key) -> Removes specified key.
# popitem() -> Removes last inserted key-value pair.

# Q6. Difference between keys(), values(), items()?
# keys()   -> All keys
# values() -> All values
# items()  -> Key-value pairs

# ==========================================================
# Quick Revision
# ==========================================================

# {}                 Create dictionary
# dict()             Create dictionary
# get()              Safe access
# update()           Add/update multiple values
# pop()              Remove key
# popitem()          Remove last pair
# del                Delete key
# clear()            Remove all items
# keys()             Get keys
# values()           Get values
# items()            Get key-value pairs
# copy()             Copy dictionary
# fromkeys()         Create dictionary from keys
# setdefault()       Add key if absent

# Important:
# ✔ Key-value pairs
# ✔ Keys must be unique
# ✔ Keys must be immutable
# ✔ Values can repeat
# ✔ Average lookup time = O(1)
```
