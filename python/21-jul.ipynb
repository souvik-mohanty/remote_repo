# Dictonary functions

```python
# keys() method returns a view object that displays a list of all the keys in the dictionary.
d2.keys()
```

```python
# values() method returns a view object that displays a list of all the values in the dictionary.
d2.values()
```

```python
#items() method returns a view object that displays a list of a dictionary's key-value tuple pairs.
d2.items()
```

```python
# update() method updates the dictionary with the specified key-value pairs.
# also used to add new key-value pairs to the dictionary.
d2.update({'job': 'Data science'})
```

```python
d2.update({'age':69})
```

```python
d2
```

```python
# popitem()-> remove last key value pair
d2.popitem()
d2
```

```python
#pop(key) -> remove certain key-value pair
d2.pop("age")
d2
```

```python
# clear() method removes all the elements from a dictionary
d2.clear()
d2
```

```python
d3={'student1': {'name': 'Alice', 'age': 20, 'city': 'New York'},
     'student2': {'name': 'Bob', 'age': 15, 'city': 'Los Angeles'},
     'student3': {'name': 'Charlie', 'age': 21, 'city': 'Chicago'}}
```

# Loop

Loop s are something which repeats code inside it again and again 

### While Loop
-> repeates statement inside it till the condition is true, when condition becomes false, loops stop

```python
i=5
while(i>0):
    print(i)
    i-=1
```

```python
str1 = "hello"
i=4
while len(str1) > i:
    print(str1[i-1])
    i-=1
    
```
