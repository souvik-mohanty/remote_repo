# Type Casting in Python

## What is Type Casting?

**Type Casting** is the process of converting one data type into another.

Python provides built-in functions for type conversion.

> **Note:** Type casting creates a new object. It does **not** modify the original value.

---

# Types of Type Casting

1. **Implicit Type Casting (Automatic)**
2. **Explicit Type Casting (Manual)**

---

# 1. Implicit Type Casting

Python automatically converts a smaller data type into a larger compatible data type.

## Example

```python id="blm8lx"
# Integer value
num1 = 10

# Float value
num2 = 5.5

# Python automatically converts int to float
result = num1 + num2

print(result)
print(type(result))

# Output:
# 15.5
# <class 'float'>
```

---

# 2. Explicit Type Casting

The programmer manually converts one data type into another using built-in functions.

## Syntax

```python id="sdb5nk"
datatype(value)
```

Example:

```python id="q4dmt0"
age = "22"

# Convert string to integer
age = int(age)

print(age)
print(type(age))

# Output:
# 22
# <class 'int'>
```

# Common Type Casting Functions

## 1. int()

Converts a value into an integer.

```python id="4xq0zf"
print(int(10.8))      # 10
print(int("25"))      # 25
print(int(True))      # 1
print(int(False))     # 0
```

---

## 2. float()

Converts a value into a floating-point number.

```python id="4g16wp"
print(float(10))
print(float("25"))
print(float(True))

# Output:
# 10.0
# 25.0
# 1.0
```

---

## 3. str()

Converts any data type into a string.

```python id="yvmejh"
age = 22

text = str(age)

print(text)
print(type(text))

# Output:
# 22
# <class 'str'>
```

---

## 4. bool()

Converts a value into a Boolean.

```python id="lmz62i"

print(bool(0))
print(bool(""))
print(bool("Python"))
print(bool([]))

# Output:
# True
# False
# False
# True
# False
# True
```

---

## 5. list()

Converts an iterable into a list.

```python id="j5k1up"

print(list(text))

# Output:
# ['P', 'y', 't', 'h', 'o', 'n']
```

---

## 6. tuple()

Converts an iterable into a tuple.

```python id="rw85s0"
numbers = [1, 2, 3]

print(tuple(numbers))

# Output:
# (1, 2, 3)
```

---

## 7. set()

Converts an iterable into a set.

```python id="hz7yq7"
numbers = [1, 2, 2, 3, 3]

print(set(numbers))

# Output:
# {1, 2, 3}
```

---

## 8. dict()

Creates a dictionary from key-value pairs.

```python id="18k5xu"

print(dict(data))

# Output:
# {'name': 'Souvik', 'age': 22}
```

````python
# ==========================================================
#              ARITHMETIC OPERATORS IN PYTHON
# ==========================================================

# ----------------------------------------------------------
# What are Arithmetic Operators?
# ----------------------------------------------------------
# Arithmetic operators are used to perform mathematical
# operations on numbers.
#
# Types of Arithmetic Operators:
# 1. +  (Addition)
# 2. -  (Subtraction)
# 3. *  (Multiplication)
# 4. /  (Division)
# 5. // (Floor Division)
# 6. %  (Modulus)
# 7. ** (Exponentiation)
#
# ----------------------------------------------------------
# Important Points
# ----------------------------------------------------------
# ✔ Arithmetic operators work with int and float values.
# ✔ Division (/) always returns a float.
# ✔ Floor Division (//) returns the quotient without decimal.
# ✔ Modulus (%) returns the remainder.
# ✔ Exponentiation (**) means power.
# ✔ Operator precedence follows BODMAS/PEMDAS.
# ==========================================================


# ==========================================================
# 1. Addition (+)
# ==========================================================

# Adds two numbers

a = 20
b = 10

print(a + b)

# Output:
# 30


# ==========================================================
# 2. Subtraction (-)
# ==========================================================

# Subtracts the second number from the first

a = 20
b = 10

print(a - b)

# Output:
# 10


# ==========================================================
# 3. Multiplication (*)
# ==========================================================

# Multiplies two numbers

a = 20
b = 10

print(a * b)

# Output:
# 200


# ==========================================================
# 4. Division (/)
# ==========================================================

# Divides one number by another.
# Always returns a float.

a = 20
b = 10

print(a / b)

# Output:
# 2.0


# Example

print(7 / 2)

# Output:
# 3.5


# ==========================================================
# 5. Floor Division (//)
# ==========================================================

# Returns only the integer quotient.
# Decimal part is discarded.

print(20 // 3)

# Output:
# 6


print(7 // 2)

# Output:
# 3


print(7.0 // 2)

# Output:
# 3.0


# ==========================================================
# 6. Modulus (%)
# ==========================================================

# Returns the remainder after division.

print(20 % 3)

# Output:
# 2


print(10 % 2)

# Output:
# 0


print(15 % 4)

# Output:
# 3


# ==========================================================
# 7. Exponentiation (**)
# ==========================================================

# Calculates power.

print(2 ** 3)

# Output:
# 8


print(5 ** 2)

# Output:
# 25


print(10 ** 0)

# Output:
# 1


# ==========================================================
# Arithmetic Operators with Variables
# ==========================================================

num1 = 15
num2 = 4

print("Addition =", num1 + num2)
print("Subtraction =", num1 - num2)
print("Multiplication =", num1 * num2)
print("Division =", num1 / num2)
print("Floor Division =", num1 // num2)
print("Modulus =", num1 % num2)
print("Power =", num1 ** num2)

# Output:
# Addition = 19
# Subtraction = 11
# Multiplication = 60
# Division = 3.75
# Floor Division = 3
# Modulus = 3
# Power = 50625


# ==========================================================
# Arithmetic Operators with Float Numbers
# ==========================================================

a = 12.5
b = 3.5

print(a + b)
print(a - b)
print(a * b)
print(a / b)

# Output:
# 16.0
# 9.0
# 43.75
# 3.5714285714285716


# ==========================================================
# Arithmetic Operators with User Input
# ==========================================================

# input() always returns a string.
# Convert the input into int before performing arithmetic.

num1 = int(input("Enter First Number: "))
num2 = int(input("Enter Second Number: "))

print("Addition =", num1 + num2)
print("Subtraction =", num1 - num2)
print("Multiplication =", num1 * num2)
print("Division =", num1 / num2)


# ==========================================================
# Operator Precedence
# ==========================================================

# Python follows BODMAS/PEMDAS.

result = 5 + 2 * 3

print(result)

# Output:
# 11


# Multiplication happens first.

result = (5 + 2) * 3

print(result)

# Output:
# 21


# ==========================================================
# Parentheses
# ==========================================================

# Parentheses have the highest priority.

print((10 + 5) * 2)

# Output:
# 30


print(10 + (5 * 2))

# Output:
# 20


# ==========================================================
# Arithmetic Assignment Example
# ==========================================================

x = 10

# Add 5
x = x + 5

print(x)

# Output:
# 15


# ==========================================================
# Negative Numbers
# ==========================================================

print(-10 + 5)

# Output:
# -5


print(-10 * -5)

# Output:
# 50


print(20 // -3)

# Output:
# -7
#
# Floor division always rounds DOWN
# toward negative infinity.


# ==========================================================
# Zero Division Error
# ==========================================================

# Division by zero is not allowed.

# print(10 / 0)

# Output:
# ZeroDivisionError


# print(10 // 0)

# Output:
# ZeroDivisionError


# print(10 % 0)

# Output:
# ZeroDivisionError


# ==========================================================
# Interview Questions
# ==========================================================

# Q1. Difference between / and // ?

# /
print(7 / 2)

# Output:
# 3.5

# //
print(7 // 2)

# Output:
# 3


# ----------------------------------------------------------

# Q2. Difference between % and // ?

# // returns Quotient
print(20 // 3)

# Output:
# 6

# % returns Remainder
print(20 % 3)

# Output:
# 2


# ----------------------------------------------------------

# Q3. What is the output?

print(2 ** 4)

# Output:
# 16


# ----------------------------------------------------------

# Q4. What is the output?

print(10 % 2)

# Output:
# 0


# ----------------------------------------------------------

# Q5. What is the output?

print(5 + 2 * 4)

# Output:
# 13


# ----------------------------------------------------------

# Q6. What is the output?

print((5 + 2) * 4)

# Output:
# 28


# ----------------------------------------------------------

# Q7. What is Exponentiation?

# ** means power.

print(3 ** 3)

# Output:
# 27


# ==========================================================
# Time Complexity
# ==========================================================

# All arithmetic operators have O(1) Time Complexity.

# +   -> O(1)
# -   -> O(1)
# *   -> O(1)
# /   -> O(1)
# //  -> O(1)
# %   -> O(1)
# **  -> O(1)


# ==========================================================
# Quick Revision
# ==========================================================

# +   Addition
# -   Subtraction
# *   Multiplication
# /   Division (Returns Float)
# //  Floor Division (Quotient)
# %   Modulus (Remainder)
# **  Exponentiation (Power)

# Important:
# ✔ / always returns float.
# ✔ // removes the decimal part by flooring.
# ✔ % returns the remainder.
# ✔ ** calculates power.
# ✔ Division by zero raises ZeroDivisionError.
# ✔ Parentheses have the highest precedence.
# ✔ Arithmetic operators work with int and float values.
# ✔ Time Complexity = O(1)
# ==========================================================
````
